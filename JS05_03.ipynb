```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Load Data (Ganti 'spam.csv' dengan nama file CSV lo yang bener)
df = pd.read_csv('spam.csv', encoding='latin-1')

# Cek nama kolom asli untuk antisipasi
# print("Kolom asli:", df.columns)

# Otomatis rename tergantung nama kolom aslinya
if 'v1' in df.columns:
    df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# Bersihin NaN
df = df.dropna(subset=['Labels', 'SMS'])

# 2. Inspeksi Data
# print("--- VALUE COUNTS ---")
# print(df['Labels'].value_counts())
# print('\n')

# Tentukan X dan y
X = df['SMS']
y = df['Labels']

# 3. Split Data Training & Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisiasi CountVectorizer & Ekstraksi Fitur
bow = CountVectorizer()
X_train = bow.fit_transform(X_train)
X_test = bow.transform(X_test)

# print(f'Jumlah Fitur (Vocabulary): {len(bow.get_feature_names_out())}')
# print(f'Dimensi data train: {X_train.shape}\n')

# 4. Training dan Evaluasi Model
mnb = MultinomialNB()
mnb.fit(X_train, y_train)

# Prediksi & Evaluasi
y_pred_train = mnb.predict(X_train)
y_pred_test = mnb.predict(X_test)

acc_train = accuracy_score(y_train, y_pred_train)
acc_test = accuracy_score(y_test, y_pred_test)

print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```
