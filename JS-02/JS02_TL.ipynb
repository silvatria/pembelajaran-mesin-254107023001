```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load dataset wbc.csv
df = pd.read_csv('wbc.csv')

# Hapus kolom id dan Unnamed: 32 (jika ada)
kolom_hapus = ['id', 'Unnamed: 32']
df = df.drop(columns=[col for col in kolom_hapus if col in df.columns])

print("Kolom setelah dihapus:", df.columns.tolist())

# Inisialisasi LabelEncoder
le = LabelEncoder()

# Fit dan transform kolom diagnosis
df['diagnosis'] = le.fit_transform(df['diagnosis'])

print("\nHasil Encoding (5 data pertama):")
print(df[['diagnosis']].head())

# Ambil semua nama kolom kecuali 'diagnosis'
kolom_numerik = df.columns.drop('diagnosis')

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Lakukan standardisasi
df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])

print("\nHasil Standardisasi (5 data pertama):")
print(df.head())
```
