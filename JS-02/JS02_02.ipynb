```python
import pandas as pd
import numpy as np

df = pd.read_csv('Titanic-Dataset.csv')

# Tampilkan 5 data teratas
df.head()

# Age - mean
df['Age'].fillna(value=df['Age'].mean())

# Cabin - "DECK"
df['Cabin'].fillna(value="DECK")

# Embarked - modus
df['Embarked'].fillna(value=df['Embarked'].mode, inplace=True)
```
